```python
import os
import numpy as np
import matplotlib.pyplot as plt
import corner

# =========================
# Paths
# =========================
posterior_folder = "/home/jsanghavi1/powerspectrumsbi/VSGsolver/outputs/11antbispectrum/samples_dir/posterior/260217-104859"
truth_file = "/home/jsanghavi1/powerspectrumsbi/VSGsolver/data/observed11z.npy"

# =========================
# Load posterior samples
# =========================
files = sorted([f for f in os.listdir(posterior_folder) if f.endswith(".npz")])

all_z = []
for f in files:
    data = np.load(os.path.join(posterior_folder, f))
    all_z.append(data["z"])  # shape (8,)

# Stack into (N_samples, 8)
pwpts = np.vstack(all_z)

print("Posterior shape:", pwpts.shape)

# =========================
# Load true parameter values
# =========================
true_vals = np.load(truth_file)

# =========================
# Labels and ranges
# =========================
labels = [f"Param {i+1}" for i in range(8)]
manual_ranges = [(-1, 1)] * 8   # change if needed

# =========================
# Select first 4 parameters
# =========================
samples_4d = pwpts[:, :4]
truths_4d = true_vals[:4]
labels_4d = labels[:4]
ranges_4d = manual_ranges[:4]

# =========================
# Corner plot
# =========================
fig = corner.corner(
    samples_4d,
    labels=labels_4d,
    truths=truths_4d,
    range=ranges_4d,
    bins=40,
    smooth=1.0,
    quantiles=[0.16, 0.5, 0.84],
    show_titles=True,
    title_fmt=".3f",
    title_kwargs={"fontsize": 11},
    truth_color="#ff4c4c",
    contour_kwargs={"linewidths": 1.5},
    fill_contours=True,
    levels=[0.68, 0.95],
)

fig.suptitle("4D Posterior Corner Plot Bispectrum", fontsize=22, y=1.02)
plt.show()
```
