```python
import corner
import matplotlib.pyplot as plt
import numpy as np
import joblib

# Load data
posterior = joblib.load("/home/jsanghavi1/powerspectrumsbi/03_gaincal/samples_posteriormaf_resamples256_200_0.01_res50_-5-2offset.joblib")
pwpts = np.array([d['pw'] for d in posterior])

# --- DEFINE YOUR CONFIGURATION HERE ---
# Replace these with your actual ground truth values
true_A = -3  
true_n = -2.5
true_vals = [true_A, true_n]

# Define manual ranges: [(min_A, max_A), (min_n, max_n)]
# Based on your title, it looks like A is (-5, -1) and n is (-4, -1)
manual_ranges = [(-5.0, -2.0), (-5.0, -2.0)] 

# ==========================================
# PLOT 1: Standard plot with True Values
# ==========================================
figure1 = corner.corner(
    pwpts,
    labels=["A", "n"],
    truths=true_vals,            # <--- Adds the red lines for true values
    truth_color="#ff4c4c",       # Optional: Custom color for truth lines
    show_titles=True
)
figure1.suptitle("Posterior with True Values", fontsize=16)
figure1.tight_layout(rect=[0, 0, 0.98, 0.98])
plt.show() # Or figure1.savefig("plot_with_truths.png")


# ==========================================
# PLOT 2: Plot with Manually Fixed Ranges
# ==========================================
figure2 = corner.corner(
    pwpts,
    labels=["A", "n"],
    range=manual_ranges,         # <--- Forces the axis limits
    truths=true_vals,            # (Optional) Keep truths in this one too
    show_titles=True
)
figure2.suptitle("Posterior with Manual Ranges", fontsize=16)
figure2.tight_layout(rect=[0, 0, 0.98, 0.98])
plt.show() # Or figure2.savefig("plot_manual_range.png")
```

```python
import corner
import matplotlib.pyplot as plt
import numpy as np
import joblib

# Load data
# Assuming pwpts is now (N, 20)
posterior = joblib.load("/home/jsanghavi1/powerspectrumsbi/03_gaincal/samples_posteriormaf_resamples128_100_0.005_res50_-5-2offset.joblib")
pwpts = np.array([d['z'] for d in posterior])

# --- DEFINE YOUR CONFIGURATION FOR 20 PARAMETERS ---
num_params = pwpts.shape[1] # Should be 20

# 1. Define Labels (First two are A and n, rest are generic)
labels = [f"p{i}" for i in range(0, num_params)]

# 2. Define Truths (Match your 20 parameters)
# Replace these with your actual 20 ground truth values
true_vals = [0.0] * (num_params) 

# 3. Define Manual Ranges (Match your 20 parameters)
# Adjust these tuples to match the expected physics for each param
manual_ranges =[(-1.0, 1.0)] * (num_params)

# =========================================================
# PLOT 1: Corner Plot for First 6 Combinations
# =========================================================
# We slice the first 6 columns: [:, :6]
figure1 = corner.corner(
    pwpts[:, :6],
    labels=labels[:6],
    truths=true_vals[:6],
    range=manual_ranges[:6],
    truth_color="#ff4c4c",
    show_titles=True,
    title_kwargs={"fontsize": 10},
    label_kwargs={"fontsize": 10}
)
figure1.suptitle("Correlations: First 6 Parameters", fontsize=16)
plt.show()

# =========================================================
# PLOT 2: 1D Distributions for All 20 Parameters
# =========================================================
# Creating a 4x5 grid for a clean look
fig, axes = plt.subplots(4, 5, figsize=(20, 15))
axes = axes.flatten()

for i in range(num_params):
    ax = axes[i]
    
    # Plot the 1D Histogram
    ax.hist(pwpts[:, i], bins=40, density=True, histtype='step', color='black', lw=1.5)
    
    # Add vertical line for the ground truth
    ax.axvline(true_vals[i], color='#ff4c4c', linestyle='--', label='Truth')
    
    # Title and Formatting
    ax.set_title(labels[i], fontsize=12)
    ax.set_xlim(manual_ranges[i])
    ax.set_yticks([]) # Hide Y-axis for cleaner look
    
    if i == 0:
        ax.legend(loc='upper right', fontsize=8)

plt.suptitle("1D Marginal Posteriors: All 20 Parameters", fontsize=22, y=1.02)
plt.tight_layout()
plt.show()
```
